Connexion a la base (lecture seule) et fonction utilitaire.

```python
import sqlite3

# Base ouverte en lecture seule
DB = "data/boutique.sqlite"
con = sqlite3.connect(f"file:{DB}?mode=ro", uri=True)

def run(sql):
    """Execute la requete et renvoie l'unique valeur revenue_cents."""
    rows = con.execute(sql).fetchall()
    assert len(rows) == 1 and len(rows[0]) == 1, "il faut 1 ligne et 1 colonne"
    return rows[0][0]

results = {}   # case_id -> (sql, valeur)
```

1. C01	period	What is the revenue excluding VAT, in euro cents, of the orders placed from 2026-03-01 00:00:00 to 2026-06-01 00:00:00?


```python
sql = """
SELECT COALESCE(SUM(o.quantity_product * p.price), 0) AS revenue_cents
FROM orders o JOIN product p ON p.product_id = o.product_id
WHERE o.date >= '2026-03-01 00:00:00' AND o.date < '2026-06-01 00:00:00';
"""
results["C01"] = (sql, run(sql))
print("C01 =", results["C01"][1])
```

2. C02	age on a reference date	What is the total revenue excluding VAT, in euro cents, generated by customers who are exactly 25 years old on the reference date 2026-09-25 (all their orders)?


```python
# Age revolu le 2026-09-25 : difference des annees, moins 1 si l'anniversaire
# (mois-jour) n'est pas encore passe a cette date.
sql = """
SELECT COALESCE(SUM(o.quantity_product * p.price), 0) AS revenue_cents
FROM orders o
JOIN product p   ON p.product_id = o.product_id
JOIN customers c ON c.customer_id = o.customer_id
WHERE CAST(strftime('%Y', '2026-09-25') AS INTEGER) - CAST(strftime('%Y', c.birth_date) AS INTEGER)
      - (strftime('%m-%d', '2026-09-25') < strftime('%m-%d', c.birth_date)) = 25;
"""
results["C02"] = (sql, run(sql))
print("C02 =", results["C02"][1])
```
